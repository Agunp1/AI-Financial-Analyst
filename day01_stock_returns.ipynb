```python
import pandas as pd

stocks = {
    "Company": ["Apple", "Microsoft", "JPMorgan", "Goldman Sachs", "BlackRock"],
    "Buy Price": [180, 400, 190, 450, 800],
    "Sell Price": [195, 430, 205, 480, 860]
}

df = pd.DataFrame(stocks)

df["Return (%)"] = (
    (df["Sell Price"] - df["Buy Price"])
    / df["Buy Price"]
) * 100

display(df.round(2))
```

```python
# Portfolio allocation
df["Weight"] = [0.25, 0.25, 0.20, 0.15, 0.15]

# Calculate each stock's contribution
df["Contribution (%)"] = df["Return (%)"] * df["Weight"]

# Calculate total portfolio return
portfolio_return = df["Contribution (%)"].sum()

display(df.round(2))
print(f"Total Portfolio Return: {portfolio_return:.2f}%")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.bar(df["Company"], df["Return (%)"])

plt.title("Stock Returns Comparison")
plt.xlabel("Company")
plt.ylabel("Return (%)")
plt.xticks(rotation=30)
plt.tight_layout()

plt.show()
```
