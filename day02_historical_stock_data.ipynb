```python
import yfinance as yf
import pandas as pd

# Download one year of Apple stock prices
apple = yf.download(
    "AAPL",
    period="1y",
    auto_adjust=True,
    progress=False
)

# Display the first five rows
display(apple.head())
```

```python
# Calculate daily returns using adjusted closing prices
apple["Daily Return"] = apple["Close"].pct_change()

# Display the latest observations
display(apple.tail())
```

```python
import matplotlib.pyplot as plt

apple["Close"].plot(
    figsize=(11, 5),
    title="Apple — Historical Stock Price",
    legend=False
)

plt.xlabel("Date")
plt.ylabel("Adjusted Closing Price ($)")
plt.grid(True)
plt.show()
```

```python
daily_returns = apple["Daily Return"].dropna()

average_daily_return = daily_returns.mean()
daily_volatility = daily_returns.std()

annualized_volatility = daily_volatility * (252 ** 0.5)

print("Average Daily Return:")
print(average_daily_return)

print("\nAnnualized Volatility:")
print(annualized_volatility)
```

```python
print(f"Average Daily Return: {average_daily_return:.3%}")
print(f"Annualized Volatility: {annualized_volatility:.2%}")
```
